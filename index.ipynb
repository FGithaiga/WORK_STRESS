# **1. EXPLORATORY DATA ANALYSIS**


```python
# Loading the dataset
df_15sn = pd.read_csv("Data/PhysiologicalSignals/physiological_signals_15sn.csv")
df_15sn
```

```python
# Displaying the shape of the DataFrame
df_15sn.shape
```

```python
# Get an overview of the df_15sn DataFrame
df_15sn.info()
```

```python
# checking missing values in df_15sn
df_15sn.isnull().sum()
```

There are no missing values in df_15sn

```python
# checking for duplicates
df_15sn.duplicated().sum()
```

There are 281 duplicates in df_15sn.

```python
# Removing duplicates if any
df_15sn.drop_duplicates(inplace=True)
```

```python
# Check number of rows before
before = df_15sn.shape[0]

# Drop duplicates
df_15 = df_15sn.drop_duplicates()

# Check number of rows after
after = df_15sn.shape[0]

print(f"Rows before: {before}")
print(f"Rows after: {after}")
print(f"Duplicates removed: {before - after}")
```

```python
# Check for remaining duplicates
duplicates_exist = df_15.duplicated().any()

if duplicates_exist:
    print("There are still duplicates.")
else:
    print("All duplicates have been removed.")
```
